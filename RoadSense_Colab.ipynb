## 4. Programmatic Prediction Examples
Run predictions directly via Python functions with custom scenarios:

```python
import os
import sys
if os.path.exists('/content/RoadSense') and not os.path.exists('models'):
    %cd /content/RoadSense
if os.getcwd() not in sys.path:
    sys.path.insert(0, os.getcwd())

from predict import load_models, predict_single, display_results

# Load models
models = load_models()

# Test Scenario 1: Extreme High-Risk Collision
high_risk_scenario = {
    'speed_limit': 70,
    'weather_conditions': 5,       # Raining + high winds
    'road_type': 6,                # Single carriageway
    'light_conditions': 6,         # Darkness - no lighting
    'road_surface_conditions': 4,  # Frost or ice
    'urban_or_rural_area': 2,      # Rural
    'junction_detail': 7,          # >4 arms junction
    'vehicle_type': 'Van / Light Goods',
    'number_of_vehicles': 3,
    'driver_age_band': 11,         # 76+ years old
    'hour': 5
}

print('Scenario 1: High-Risk Collision')
results_1 = predict_single(models, high_risk_scenario)
display_results(results_1)

# Test Scenario 2: Low-Risk Commute Collision
low_risk_scenario = {
    'speed_limit': 20,
    'weather_conditions': 1,       # Fine
    'road_type': 1,                # Roundabout
    'light_conditions': 1,         # Daylight
    'road_surface_conditions': 1,  # Dry
    'urban_or_rural_area': 1,      # Urban
    'junction_detail': 1,
    'vehicle_type': 'Car',
    'number_of_vehicles': 2,
    'hour': 14
}

print('\nScenario 2: Low-Risk Collision')
results_2 = predict_single(models, low_risk_scenario)
display_results(results_2)
```
